```python
# a. Dada una lista de temperaturas en Celsius, convertirlas a Fahrenheit.
temp_celsiuis = [10, 20, 30, 40, 50]
temp_fahrenheit = list(map(lambda celsiuis: (celsiuis * 9/5) + 32, temp_celsiuis))
print(temp_fahrenheit)
```

```python
# b. Dadas dos listas con la misma cantidad de elementos, generar una nueva sumando cada elementos (primer elemento de lista 1 + primer elemento lista 2).
lista_1 = [10, 20, 30, 40, 50]
lista_2 = [100, 200, 300, 400, 500]
nueva_lista = list(map(lambda x, y: x + y, lista_1, lista_2))
print(nueva_lista)
```

```python
# c. Dada una lista de radios de circulos, generar la lista de sus areas.
radios = [10, 20, 30, 40, 50]
areas = list(map(lambda radio: 3.14 * (radio ** 2), radios))
print(areas)
```

```python
# d. Dada un diccionario, generar una lista de tuplas (nombre, valor+10%)
inventario = {"pan": 1000, "leche": 2500, "café": 5000}
tuplas = list(map(lambda item: (item[0], item[1]*(1+0.1)), inventario.items()))
print(tuplas)
```

```python
# e. Dada una matriz, multiplicar todos los elementos por 10.
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
matriz_10 = list(map(lambda fila: list(map(lambda columna: columna * 10, fila)), matriz))
print(matriz_10)
```
