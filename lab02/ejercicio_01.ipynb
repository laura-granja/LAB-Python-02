```python
multiplo_3 = lambda x: x % 3 == 0
cubo = lambda x: x**3
producto = lambda x, y: x * y
mayor = lambda x, y: max(x, y)
empieza_a = lambda palabra: palabra[0] == "a" or palabra[0] == "A"
celsuis_a_farenheit = lambda celsuis: (celsuis * 9/5) + 32
print(multiplo_3 (3))
print(multiplo_3 (2))
print(cubo (3))
print(producto(4, 5))
print(mayor(15, 5))
print(empieza_a("ana"))
print(empieza_a("pedro"))
print(empieza_a("Ana"))
print(celsuis_a_farenheit(35))
```

```python
operaciones = [lambda x: 2 * x, lambda x: x + 10, lambda x: x**2]
x = int(input("Ingrese un número: "))
for operacion in operaciones:
    print(operacion(x))
```
