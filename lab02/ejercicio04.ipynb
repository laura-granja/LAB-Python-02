# Ejercicio 4 - Filter


```python
# a. Dada una lista de palabras, filtrar solo aquellas con mas de 4 letras.
palabras = ["hola", "mundo", "programacion", "python"]
mas_4_letras = list(filter(lambda palabra: len(palabra) > 4, palabras))
print(mas_4_letras)
```

```python
# b. Dada una lista filtrar los elementos que son nulos (None).
nulos = [1, None, 2, None, 3, 4, None]
sin_nulos = list(filter(lambda elemento: elemento is not None, nulos))
print(sin_nulos)
```

```python
# c. Dada una lista de palabras, filtrar aquellas que empiezan por una vocal
vocales = ["a", "e", "i", "o", "u"]
palabras = ["hola", "mundo", "Oso", "programacion", "python", "arbol", "elefante"]
palabras_vocal = list(filter(lambda palabra: palabra[0].lower() in vocales, palabras))
print(palabras_vocal)
```

```python
# d. Dada una lista de palabras, filtrar aquellas que son palindromos
palabras = ["oso", "hola", "radar", "reconocer", "roma", "python"]
palindromos = list(filter(lambda palabra: palabra == palabra[::-1], palabras))
print(palindromos)
```

```python
# e. Dada una lista de numeros, filtrar los que terminan en 5.
numeros = [10, 15, 20, 25, 30, 35, 40, 45, 50]
terminan_5 = list(filter(lambda numero: numero % 10 == 5, numeros))
print(terminan_5)
```

```python
# f. Dada una lista de diccionarios, filtrar los productos con precio mayor a 100.
productos = [
    {"nombre": "Teclado", "precio": 80},
    {"nombre": "Mouse", "precio": 40},
    {"nombre": "Monitor", "precio": 300},
    {"nombre": "Webcam", "precio": 150}
]
precio_mayor_100 = list(filter(lambda producto: producto["precio"] > 100, productos))
print(precio_mayor_100)
```

# Ejercicio 4 - Reduce


```python
from functools import reduce
# a. Dada una lista de numeros, calcula la multiplicación de estos.
numeros = [1, 2, 3, 4, 5]
multiplicacion = reduce(lambda x, y: x * y, numeros)
print(multiplicacion)
```

```python
# b. Dada una lista de palabras, concetenalas en una sola.
concanetacion = reduce(lambda palabra_1, palabra_2: palabra_1 + palabra_2, palabras)
print(concanetacion)
```

```python
# c. Dada una lista, hallar el mayor de todos, define una función auxiliar que dado dos numeros retorne el mayor
def mayor(x, y):
  return x if x > y else y
numeros = [10, 5, 20, 15, 30]
mayor_numero = reduce(mayor, numeros)
print(mayor_numero)
```

# Ejercicio 4 - Sorted


```python
# a. Ordenas una lista de palabras alfabeticamente.
palabras = ["manzana", "banana", "cereza", "naranja", "uva"]
palabras_ordenadas = sorted(palabras)
print(palabras_ordenadas)
```

```python
# b. Ordenar una lista de tuplas por el primer elemento.
tuplas = [(2, "b"), (1, "a"), (3, "c")]
tuplas_ordenadas = sorted(tuplas, key= lambda tupla: tupla[0])
print(tuplas_ordenadas)
```

```python
# c. Dada una lista de diccionarios, ordenar por precio (mayor a menor), y luego por nombre
productos = [
      ("Camisa", 30),
      ("Pantalón", 55),
      ("Medias", 10),
      ("Chaqueta", 80)
  ]
productos_ordenados = sorted(productos, key= lambda producto: (-producto[1], producto[0]))
print(productos_ordenados)
```

```python
# d. Dada una lista de palabras, ordenarlas según su longitud.
palabras_ordenadas_longitud = sorted(palabras, key= lambda palabra: len(palabra))
print(palabras_ordenadas_longitud)
```

```python
# e. Dada una lista de numeros, ordenarlos de tal forma que aparezcan primero los pares y luego los impares.
numeros = [1, 2, 3, 8, 19, 20]
pares_impares = sorted(numeros, key= lambda x: x % 2 != 0)
print(pares_impares)
```

```python
# f. Convertir una palabra a un diccinario que cuente la aparicion de cada letra, ordenar dicho diccionario según la frecuencia de aparicion.
diccionario = {}
palabra = "programacion"
for letra in palabra:
  if letra in diccionario:
    diccionario[letra] += 1
  else:
    diccionario[letra] = 1
ordenado = sorted(diccionario, key= lambda letra: diccionario[letra])
print(ordenado)
```

```python
# g. Dada una lista de tuplas, ordenar la primero por el departamento del empleado y luego por la edad.
empleados = [
      ("Maria", "Ventas", 30),
      ("Luis", "Tecnología", 25),
      ("Ana", "Ventas", 25),
      ("Pedro", "Tecnología", 28)
    ]
empleados_ordenados = sorted(empleados, key= lambda empleado: (empleado[1], empleado[2]))
print(empleados_ordenados)
```
